```python

import sqlite3
import pandas as pd

conn = sqlite3.connect(":memory:")

conn.execute("""
CREATE TABLE Employees (
    id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary INTEGER,
    age INTEGER
)
""")

conn.executemany("""
INSERT INTO Employees VALUES (?, ?, ?, ?, ?)
""", [
    (1, "Arun", "IT", 60000, 23),
    (2, "Bala", "HR", 45000, 25),
    (3, "Charan", "IT", 75000, 24),
    (4, "Divya", "Sales", 50000, 22),
    (5, "Ezhil", "HR", 45000, 26),
    (6, "Farah", "IT", 80000, 27),
    (7, "Gokul", "Sales", 55000, 23)
])

conn.commit()

print("Database created successfully!")
```

```python

pd.read_sql_query("SELECT * FROM Employees", conn)
```

```python

pd.read_sql_query("""
SELECT name, salary
FROM Employees
WHERE salary > 50000
""", conn)
```

```python

pd.read_sql_query("""
SELECT name, salary
FROM Employees
ORDER BY salary DESC
""", conn)
```

```python

pd.read_sql_query("""
SELECT MAX(salary) AS highest_salary
FROM Employees
""", conn)
```

```python

pd.read_sql_query("""
SELECT department, COUNT(*) AS employee_count
FROM Employees
GROUP BY department
""", conn)
```

```python

pd.read_sql_query("""
SELECT MAX(salary) AS second_highest_salary
FROM Employees
WHERE salary < (SELECT MAX(salary) FROM Employees)
""", conn)
```

```python

pd.read_sql_query("""
SELECT department, AVG(salary) AS average_salary
FROM Employees
GROUP BY department
""", conn)
```

```python

pd.read_sql_query("""
SELECT salary, COUNT(*) AS frequency
FROM Employees
GROUP BY salary
HAVING COUNT(*) > 1
""", conn)
```

```python

pd.read_sql_query("""
SELECT department, AVG(salary) AS average_salary
FROM Employees
GROUP BY department
""", conn)
```

```python

pd.read_sql_query("""
SELECT name, salary
FROM Employees
WHERE salary > (SELECT AVG(salary) FROM Employees)
""", conn)
```

```python

pd.read_sql_query("""
SELECT name, salary,
       DENSE_RANK() OVER (ORDER BY salary DESC) AS salary_rank
FROM Employees
""", conn)
```

```python
pd.read_sql_query("""
SELECT *
FROM Employees
WHERE name LIKE 'A%'
""", conn)
```

```python
pd.read_sql_query("""
SELECT name, salary
FROM Employees
WHERE salary BETWEEN 50000 AND 75000
""", conn)
```

```python
pd.read_sql_query("""
SELECT department, SUM(salary) AS total_salary
FROM Employees
GROUP BY department
""", conn)
```

```python
pd.read_sql_query("""
SELECT name, age
FROM Employees
WHERE age > (SELECT AVG(age) FROM Employees)
""", conn)
```
